```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')

titanic.head()
```

```python
titanic.info()
```

```python
titanic.describe()
```

```python
titanic.isnull().sum()
```

```python
numeric_data = titanic.select_dtypes(include=['number'])

plt.figure(figsize=(10,8))

sns.heatmap(numeric_data.corr(), annot=True, cmap='coolwarm')

plt.title("Correlation Heatmap")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='survived', data=titanic)

plt.title("Survival Count")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='sex', hue='survived', data=titanic)

plt.title("Survival by Gender")

plt.show()
```

```python
plt.figure(figsize=(6,4))

sns.countplot(x='class', data=titanic)

plt.title("Passenger Class Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(titanic['age'], bins=30, kde=True)

plt.title("Age Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.histplot(titanic['fare'], bins=30, kde=True)

plt.title("Fare Distribution")

plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.boxplot(x=titanic['fare'])

plt.title("Fare Boxplot")

plt.show()
```

```python
sns.pairplot(
    titanic[['survived','age','fare','pclass']]
)

plt.show()
```

```python
# Key Insights

1. The Titanic dataset contains passenger information such as age, gender, class, fare, and survival status.

2. The dataset contains missing values, primarily in the Age, Cabin, and Embarked columns.

3. Female passengers had a significantly higher survival rate than male passengers.

4. Most passengers traveled in Third Class.

5. The majority of passengers were young adults.

6. Passenger fares varied greatly, with several high-value outliers.

7. There is a relationship between passenger class and fare, where higher-class passengers generally paid higher fares.

8. The correlation heatmap indicates moderate relationships among numerical features.

9. Visualizations helped identify trends that are not immediately obvious from raw data.

10. Exploratory Data Analysis provides valuable insights before building machine learning models.
```
